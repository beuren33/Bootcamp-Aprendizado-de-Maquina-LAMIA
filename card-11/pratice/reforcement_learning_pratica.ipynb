```python
import numpy as np
import gymnasium as gym
from IPython.display import clear_output
from time import sleep
```

```python
np.random.seed(2)
env = gym.make("FrozenLake-v1", is_slippery=False)
# cria o ambiente do minijogo na variavel
acoes = env.action_space.n
# pega quantas acoes sao possiveis no jogo
q = np.zeros((env.observation_space.n, acoes))
# cria uma matrz de zeros com o numero de estados possiveis x numero de ações
```

```python
taxa_aprendizado = 0.8
# quanto o agente aprende a cada rodada
fat =0.95
# quanto as recompensas futuram importam
ex = 1.0
#quanto o agente explora o mapa
epochs = 2000
# quantas rodadas o loop do jogo tem
```

```python

for ep in range(epochs):
    state, i = env.reset()
    # renicia o ambiente
    done = False
    while not done:
        action = np.random.randint(acoes) if np.random.rand() < ex else np.argmax(q[state])
        # gerado um numero entre 0 e 1, se for menor que ex escolhe uma acao aleatoria, se não pega uma acao da tabela q
        next_state, recompensa, terminated, truncated, i = env.step(action)
        # apos o agente dar o passo recebe o proximo passo, sua recompensa, se terminou o jogo(caiu em um buraco), se atingiu o limite de passos
        done = terminated or truncated
        # se perdeu e acabou os passos vai como False, se nao vai true e contnuua
        q[state, action] += taxa_aprendizado * (recompensa + fat * np.max(q[next_state]) - q[state, action])
        # formula do q-learning, calcula quao bom foi esse passo
        state = next_state
        #proximo passo
    ex = max(0.1, ex * 0.995)  
    # reduz a exploracao aos poucos
```
