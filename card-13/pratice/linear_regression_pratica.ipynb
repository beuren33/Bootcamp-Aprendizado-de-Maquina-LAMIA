```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
```

```python
dados_speed = np.random.normal(10, 2, 10000)
dados_compra = 100 -(dados_speed + np.random.normal(0,1, 10000)) *3
```

```python
plt.scatter(dados_speed, dados_compra)
```

```python
slope, intercept, r_value, p_value, std_err = stats.linregress(dados_speed, dados_compra)
# usa formula de regressao linear do scipy

r_squared = r_value**2
r_squared
```

```python
def predict(x):
    return slope *x +intercept

line = predict(dados_speed)
plt.scatter(dados_speed, dados_compra)
plt.plot(dados_speed, line, c='y')
```

```python
dados_speed2 = np.random.normal(10, 5, 10000)
dados_compra2 = 100 -(dados_speed2 + np.random.normal(0,2, 10000)) *3
```

```python
line_e = predict(dados_speed2)
plt.scatter(dados_speed2, dados_compra2)
plt.plot(dados_speed2, line_e, c='y')
```
