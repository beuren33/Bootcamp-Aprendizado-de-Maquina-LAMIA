```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import statsmodels.api as sm
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder
```

```python
df = pd.read_excel('http://cdn.sundog-soft.com/Udemy/DataScience/cars.xls')
```

```python
# fiz o teste com uma feature de string, por isso utilizeei o label encoder
# ja tinha usado o label em outros projetos
# ele vai transformar as strings da coluna MAke em numeros categoricos
# nesse caso tem a chance do modelo achar que por exemplo, 2 é maior que 1, mas isso é so a categoria do carro
# o R-squared de 0.75 um numero ate que bom mas nao se compara ao exemlo do professor. Por conta da transformação da feature pode ter alterado o R-squared
label = LabelEncoder()

df['Model_label'] = label.fit_transform(df['Model'])

print(df['Model_label'])
```

```python
scale = StandardScaler()
# pega o objeto standardScaler

X = df[['Mileage','Cylinder','Doors', 'Model_label']]
# pega 5 features para tentar prever o valor do preço
y = df['Price']
# atribui a y, a target_feature

X[['Mileage','Cylinder','Doors', 'Model_label']] = scale.fit_transform(X[['Mileage','Cylinder','Doors', 'Model_label']])
# normaliza os dados para -1 e 1

print(X)

x_train = sm.OLS(y,X).fit()
# treina um modelo de regressao 

x_train.summary()
```

```python

# agrupa por feature e faz as medias
# tudo em relação ao preço
print(y.groupby(df['Cylinder']).mean())
print(y.groupby(df['Doors']).mean())
print(y.groupby(df['Model_label']).mean())
```

```python
#esta celula cria uma predição com as features de caracteristicas e faz uma predição
code_model = label.transform(['Cavalier'])[0]
input = scale.transform([[45000,8,4, code_model]])

predicted = x_train.predict(input)
print(predicted)
```
