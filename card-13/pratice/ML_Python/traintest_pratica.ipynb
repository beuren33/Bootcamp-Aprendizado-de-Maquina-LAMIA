```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import r2_score
```

```python
np.random.seed(2)
pageSpeeds = np.random.normal(3,1.0,100)
purchaseAmount = np.random.normal(50,30,100) / pageSpeeds

plt.scatter(pageSpeeds, purchaseAmount)
```

```python
trainX = pageSpeeds[:80]
# primeiros 800 numeros
testX = pageSpeeds[80:]
#pga o resto dos numeros 

trainY = purchaseAmount[:80]
testY = purchaseAmount[80:]
# mesma coisa para os dados de teste
```

```python
x = np.array(trainX)# transforma em array numpy
y=np.array(trainY)

testx= np.array(testX) # coloca os dados de teste em array numpy
testy =np.array(testY)

list_pol = []
for i in range(1,10):
    p =np.poly1d(np.polyfit(x,y,i))#treina o modelo com 9 polinomios diferentes
    list_pol.append(r2_score(testy, p(testx)))# pega o r2 de cada treino
```

```python
list_pol
```

```python
barras = plt.bar(range(1,10), list_pol)
# cria um grafico de barras para ver os r2 scores de cada test de polinomio
plt.bar_label(barras, fmt="%.2f")
plt.show()
```

```python
best_pol =max(list_pol)
# pega o melhor polinomio(maior r2)
best_pol
```
