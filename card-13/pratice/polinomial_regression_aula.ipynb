```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import r2_score
```

```python
np.random.seed(33)
dadosS = np.random.normal(3,1,1000)
dadosP = np.random.normal(50,10,1000) / dadosS
# aumentando os dados gerados, nao correlaconados

plt.scatter(dadosS, dadosP)
```

```python
x = np.array(dadosS)
y = np.array(dadosP)
# colocando em arrays para usar em treino
```

```python
list_scores = []
grs = []
for i in range(1,16):
    p =np.poly1d(np.polyfit(x,y,i))
    list_scores.append(r2_score(y,p(x)))
    grs.append(i)


# for que testa 20 polinomios diferentes
```

```python
print(list_scores)
```

```python
plt.bar(grs, list_scores, color="blue")
plt.xlabel("Grau do polinomio")
plt.ylabel("R²")
plt.ylim(0, 1)
plt.grid(axis="y", alpha=0.3)
plt.show()
```
