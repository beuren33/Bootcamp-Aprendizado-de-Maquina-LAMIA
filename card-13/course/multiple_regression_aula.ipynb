```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import statsmodels.api as sm
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_excel('http://cdn.sundog-soft.com/Udemy/DataScience/cars.xls')
```

```python
df1 = df[['Mileage','Price']]
# seleciona as duas colunas
bins = np.arange(0,50000,10000)
# cria 5 faixas com 10000 cada
groups = df1.groupby(pd.cut(df1['Mileage'], bins)).mean()
# separa os dados de Mileage pelos grupos nas faixas criadas bins(agrupa os dados pelas faixas bins), depois faz a media
print(groups.head())
# 5 primeiras linhas normalizadas
groups['Price'].plot.line()
```

```python
scale = StandardScaler()
# pega o objeto standardScaler

X = df[['Mileage','Cylinder','Doors']]
# atrivui a X as 3 features com caracteristicas
y = df['Price']
# atribui a y a target_feature

X[['Mileage', 'Cylinder', 'Doors']] = scale.fit_transform(X[['Mileage', 'Cylinder', 'Doors']])
# processa os dados das features de caracteristicas

print(X)

est = sm.OLS(y,X).fit()
# treina um modelo de regressao

est.summary()
```

```python
y.groupby(df.Doors).mean()
# agrupa o preço pela quantidade de portas
```

```python
#esta celula cria uma predição com as features de caracteristicas e faz uma predição
scaled = scale.transform([[45000,8,4]])
print(scaled)
predicted = est.predict(scaled[0])
print(predicted)
```
