```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
```

```python
pageSpeeds = np.random.normal(3.0, 1.0,1000)
# geração de 100 numeros aleatórios com media de 3 e desvio padrao de 1
purchaseAmount = 100 - (pageSpeeds + np.random.normal(0,0.1,1000))*3
#formula para correlacionar pagina rapida e compra
# gera dados correlacionados com pagespeeds
# logica: paginas mais rapidas = mais compras, paginas lentas = menos compras

plt.scatter(pageSpeeds, purchaseAmount)
```

```python
slope, intercept, r_value, p_value, std_err = stats.linregress(pageSpeeds, purchaseAmount)
# calcula a regressao linear
# slope: inclinação da reta, ou seja, quanto uma variavel muda quando a outra muda
# intercept: onde a linha cruza o eixo y
# r_value: coeficiente de relação
# p_value: foi aleatorio ou nao a metrica
# std_err: erro de variancia
```

```python
r_value**2
# calcula o R² r-squared
```

```python
def predict(x):
    return slope *x +intercept
# faz uma predição predição
fitline = predict(pageSpeeds)

plt.scatter(pageSpeeds, purchaseAmount)
plt.plot(pageSpeeds, fitline, c='r')
plt.show()
```

### aumentar a variação aleatória dos dados de teste e observar o impacto no R

```python
pageSpeeds = np.random.normal(3.0, 3.0,1000)

purchaseAmount = 100 - (pageSpeeds + np.random.normal(0,1,1000))*3


plt.scatter(pageSpeeds, purchaseAmount)
```
