```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
import xgboost as xgb
from sklearn.metrics import accuracy_score
```

```python
iris = load_iris()
# carrega o dataset de flores iris
numSamples, numFeatures = iris.data.shape
# separa o numero de fores e features
print(numSamples)
print(numFeatures)
print(list(iris.target_names))
```

```python
x_train, x_test, y_train, y_test = train_test_split(iris.data, iris.target, test_size=0.2, random_state=0)
# separa os dados em dados de treino e teste
```

```python
train = xgb.DMatrix(x_train, label=y_train)
test = xgb.DMatrix(x_test, label=y_test)
# empacota os dados de treino e teste para otimização do modelo
```

```python
param = {
    'max_depth': 4,
    'eta':0.3,
    'objective':'multi:softmax',
    'num_class': 3
}
# parametros passados por dicionario
epochs = 10
```

```python
model = xgb.train(param,train, epochs)
# treino do modelo com os parametros colcoados acima
```

```python
predictions = model.predict(test)
# faz predicoes com a matriz de testes
```

```python
print(predictions)
```

```python
accuracy_score(y_test, predictions)
# mede a acuraia das predicoes compara aos dados de tests
```
