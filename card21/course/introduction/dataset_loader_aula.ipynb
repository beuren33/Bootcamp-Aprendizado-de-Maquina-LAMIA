```python
from torch.utils.data import Dataset, DataLoader
import torch
import torch.nn as nn
```

```python
x = [[1, 2], [3, 4], [5, 6], [7, 8]]
y = [[3], [7], [11], [15]]
```

```python
x = torch.tensor(x, dtype=torch.float32)
y = torch.tensor(y, dtype=torch.float32)
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
x = x.to(device)
y = y.to(device)
```

```python
class MyDataset(Dataset):
    # cria um dataset proprio para usar o DataLoader
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __len__(self):
        return len(self.x)

    def __getitem__(self, idx):
        return self.x[idx], self.y[idx]
```

```python
ds = MyDataset(x, y)
```

```python
dl = DataLoader(ds, batch_size=2, shuffle=True)# agrupa as amostras em grupo de 2 e embaralha
```

```python
for x,y in dl:
    # mostra como fica os dados depois de passar pelo DataLoader
    print(x, y)
```

```python
class Rede(nn.Module):
    def __init__(self):
        super(Rede, self).__init__()
        self.layer1 = nn.Linear(2, 8)# camada de entrada 
        self.activation = nn.ReLU()# camda de ativação
        self.layer2 = nn.Linear(8, 1)# camada de saída

    def forward(self, x):
        x = self.layer1(x)
        x = self.activation(x)
        x = self.layer2(x)
        return x
```

```python
model = Rede()
loss = nn.MSELoss()# define a funcao de erro
optimizer = torch.optim.SGD(model.parameters(), lr=0.01)# oimizador que vai ajustar os pesos da rede
```

```python
losses = []
for _ in range(100):
    for data in dl:
        optimizer.zero_grad()# zera o gradient
        x, y = data # pega o dado de treino e seu valor real
        l = loss(model(x), y)#aplica a funcoa de loss ao resultado da rede e o valor real
        l.backward() # calcula o gradient
        optimizer.step() # atualiza os pesos da rede
        losses.append(l.item())
```
