```python
import torch
```

```python
x=torch.tensor([[1,2,3],[4,5,6]])
# cria um tensor de duas dimensoes
```

```python
print(x.shape)
# mostrab sua dimensao
```

```python
x*10
# multiplica por 10
```

```python
x.add(10)
# soma 10 para cada numero dentro do tensorflow
```

```python
x.view(3,2)
```

```python
a = torch.ones((2,1,10))
# cria um tensor de 3 dimensoes preenchidos com 1
a
```

```python
print(a.shape)
squeezed = a.squeeze(1)
# muda a dimensao do tensor removendo a dimensao 1
print(squeezed.shape)   
```

```python
a = torch.ones((2,2,10))
print(a.shape)
squeezed = a.squeeze(1)
print(squeezed.shape)
# nao muda a dimensao do tensor pois nao existe dimensao 1
```

```python
a = torch.tensor([[1,2,3],[4,5,6]])
print(a.shape)
unsqueezed = a.unsqueeze(2)
print(unsqueezed.shape)
```

```python
a=torch.ones(2,3)
```

```python
print(a[None].shape)
print(a[:,None].shape)
print(a[:,:,None].shape)
# manipulação de indices para adicionar dimensões ao tensor
```

```python
x = torch.tensor([[1,2,3,4],[5,6,7,8]])
# cria uma matriz 2x2
y= torch.tensor([[1,2,3],
[2,3,4],
[4,5,6],
[7,8,9]])
# cria uma matriz de 4x3
```

```python
torch.matmul(x,y)
# multiplica os dois tensors(matrizes) retornando uma matriz 2x3
```

```python
x = torch.randn(2,3,2)
z=torch.cat((x,x),dim=1)
print(x.shape,z.shape)
```

```python
x = torch.randn(2,30,20)
z=torch.permute(x,(0,2,1))
# muda a ordem das dimensoes do tensor, nesse caso a dimensao 1 passa a ser 2 e a 2 passa a ser 1
print(x.shape,z.shape)
```
