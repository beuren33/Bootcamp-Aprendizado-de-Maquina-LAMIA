```python
from ucimlrepo import fetch_ucirepo
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split, cross_val_score
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras import utils as np_utils
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import confusion_matrix
from scikeras.wrappers import KerasClassifier
from sklearn.model_selection import GridSearchCV
```

```python
wine = fetch_ucirepo(id=109 )
x = wine.data.features
y = wine.data.targets
# pega o dataset de qulidade de vinho no uci
```

```python
le = LabelEncoder()
y = le.fit_transform(y.values)
# transformo os valores strings em numeros
y
```

```python
y = np_utils.to_categorical(y)
# coloc em categorias
y
```

```python
x.columns.value_counts().sum()
```

```python
def rede(neurons):
    rede = Sequential([
        tf.keras.layers.InputLayer(shape=(x.columns.value_counts().sum(),)),
        # 13 features na camada de entrada
        tf.keras.layers.Dense(units=neurons,activation='relu'),
        # primeira camada densa
        tf.keras.layers.Dense(units=neurons,activation='relu'),
        # segunda camada densa
        tf.keras.layers.Dense(units=3,activation='softmax'),
        # saida com 3 classes, softmax retorna a probabilidade de cada uma delas
    ])
    rede.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['categorical_accuracy'])
    return rede
```

```python
rede = KerasClassifier(model = rede)
```

```python
params ={
    'batch_size':[10],
    'epochs':[200],
    'model__neurons': [8,10,12]
}# parametros testados
```

```python
gs = GridSearchCV(estimator=rede, param_grid=params,scoring='accuracy',cv=5)
# testa todos os paramtros, com cross validation de 5 para 5 partes do dataset
```

```python
gs = gs.fit(x,y)
# treina o modelo
```

```python
acc=gs.best_score_
print(acc)
# melhor acuracia
```
