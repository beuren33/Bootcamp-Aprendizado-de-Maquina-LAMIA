```python
import pandas as pd
import tensorflow as tf
from sklearn.model_selection import train_test_split
from tensorflow.keras.models import Sequential
```

```python
x = pd.read_csv('/home/beuren/Documentos/LAMIA-Bootcamp/card-14/data/classificação binária/entradas_breast.csv')
y= pd.read_csv('/home/beuren/Documentos/LAMIA-Bootcamp/card-14/data/classificação binária/saidas_breast.csv')
# carregas os dados das features e a feature target
```

```python
x_train, x_test,y_train,y_test = train_test_split(x,y,test_size=0.25)
# afz separação dos dados em teste e treino
```

```python
(30 +1) /2
#faz o calculo para saber o numero ideal de neuronios na camada densa
```

```python
rede_neural = Sequential([
    tf.keras.layers.InputLayer(shape=(30,)),
    # recebe uma tupla com o total de neuronios de entrada, 30 por onta das 30 features
    tf.keras.layers.Dense(units=16, activation='relu', kernel_initializer='random_uniform'),
    # define a quantidade de neuronios na camada densa com funcao de ativação relu
    tf.keras.layers.Dense(units=16, activation='relu', kernel_initializer='random_uniform'),
    # duas camadas ocultas na rede
    tf.keras.layers.Dense(units=1, activation='sigmoid')
    # escolhe a saida como 1 pois e uma classificação binaria entao usa somente uma saida com a funcao sigmoide para retornar a probabilidade
])
```

```python
rede_neural.summary()
```

```python
otimizador = tf.keras.optimizers.Adam(learning_rate=0.001, clipvalue=0.5)
```

```python
rede_neural.compile(optimizer=otimizador, loss= 'binary_crossentropy', metrics=['binary_accuracy'])
# parametros utilizados, o otimizador adam, o ajustados de pesos, e como medir os resultados do trein
```

```python
rede_neural.fit(x_train,y_train, batch_size=10, epochs=100)
# treina o modelo com 100 registros por vez, e 100 vezes por todos os dados
```
