```python
import pandas as pd
import tensorflow as tf
import sklearn
import scikeras
```

```python
from scikeras.wrappers import KerasClassifier
from sklearn.model_selection import cross_val_score
from tensorflow.keras.models import Sequential
from tensorflow.keras import backend as k
```

```python
x = pd.read_csv('/home/beuren/Documentos/LAMIA-Bootcamp/card-14/data/classificação binária/entradas_breast.csv')
y= pd.read_csv('/home/beuren/Documentos/LAMIA-Bootcamp/card-14/data/classificação binária/saidas_breast.csv')
```

```python
def criar_rede():
    k.clear_session()
    rede_neural = Sequential([
        tf.keras.layers.InputLayer(shape=(30,)),
        # rentrada com 30 neuronios
        tf.keras.layers.Dense(units=16, activation='relu', kernel_initializer='random_uniform'),
        # primeira camada oculta
        tf.keras.layers.Dense(units=16, activation='relu', kernel_initializer='random_uniform'),
        # segunda camada oculta
        tf.keras.layers.Dense(units=1, activation='sigmoid')
        # um neuronio de saida
    ])
    otimizador = tf.keras.optimizers.Adam(learning_rate= 0.001, clipvalue=0.5)
    rede_neural.compile(optimizer = otimizador, loss='binary_crossentropy', metrics = ['binary_accuracy'])
    return rede_neural
```

```python
rede_neural = KerasClassifier(model = criar_rede, batch_size=10, epochs=100)
```

```python
resultados = cross_val_score(estimator=rede_neural, X=x, y=y, cv=10, scoring='accuracy')
```
