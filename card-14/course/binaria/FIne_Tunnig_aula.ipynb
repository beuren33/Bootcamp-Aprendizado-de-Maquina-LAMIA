```python
import pandas as pd
import tensorflow as tf
import sklearn
import scikeras
```

```python
from scikeras.wrappers import KerasClassifier
from tensorflow.keras import backend as k
from tensorflow.keras.models import Sequential
from sklearn.model_selection import GridSearchCV
```

```python
x = pd.read_csv('/home/beuren/Documentos/LAMIA-Bootcamp/card-14/data/classificação binária/entradas_breast.csv')
y= pd.read_csv('/home/beuren/Documentos/LAMIA-Bootcamp/card-14/data/classificação binária/saidas_breast.csv')
```

```python
def criar_rede(optimizer, loss, kernel_initializer, activation, neurons):
  k.clear_session()
  rede_neural = Sequential([
      tf.keras.layers.InputLayer(shape=(30,)),
      tf.keras.layers.Dense(units=neurons, activation=activation, kernel_initializer=kernel_initializer),
      tf.keras.layers.Dropout(rate = 0.2),
      tf.keras.layers.Dense(units=neurons, activation=activation, kernel_initializer=kernel_initializer),
      tf.keras.layers.Dropout(rate = 0.2),
      tf.keras.layers.Dense(units=1, activation = 'sigmoid')])
  rede_neural.compile(optimizer = optimizer, loss = loss, metrics = ['binary_accuracy'])
  return rede_neural

# cria a ee com 3 camadas ocultas, todas com dropout para overfitting, mesmo problema binario
```

```python
rede_neural = KerasClassifier(model = criar_rede)
```

```python
parametros = {
    'batch_size': [10, 30],
    'epochs': [50],
    'model__optimizer': ['adam'],
    'model__loss': ['binary_crossentropy'],
    'model__kernel_initializer': ['random_uniform', 'normal'],
    'model__activation': ['relu'],
    'model__neurons': [16]
}# define os paremetros que serao testados no processo de fine tunnig
```

```python
parametros
```

```python
2 * 1 * 1 * 1 * 2 * 1 * 1 * 5
# total de combinaçoes que serao feitas pelo grid search
```

```python
grid_search = GridSearchCV(estimator = rede_neural, param_grid = parametros,scoring = 'accuracy', cv = 5)
# o grid search,testa diversas combinações de parametros para achar o melhor modelo
```

```python
grid_search = grid_search.fit(x, y)
```

```python
print(grid_search)
```

```python
melhores_parametros = grid_search.best_params_
# pega somente os melhores parametros do fine tunning
print(melhores_parametros)
```

```python
melhor_precisao = grid_search.best_score_
# a melhor acoracia encontrada
print(melhor_precisao)
```
