```python
import pandas as pd 
import numpy as np 
import sklearn
import tensorflow as tf 
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras import utils as np_utils
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix
```

```python
base = pd.read_csv('/home/beuren/Documentos/LAMIA-Bootcamp/card-14/data/classificação mais classes/iris.csv')
base
```

```python
x = base.iloc[:, 0:4].values
# pega as features previsoras
x
```

```python
y = base.iloc[:,4].values
# pega os dados rotulados(targets)
y
```

```python
label_encoder = LabelEncoder()
y = label_encoder.fit_transform(y)
# transforma as strings em numeros para o modelo entender
y
```

```python
y = np_utils.to_categorical(y)
# transforma em categorias, uma matriz de tres colunas para cada tipo de flor
y
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.25)
```

```python
x_train.shape, x_test.shape
```

```python
y_train.shape, y_test.shape
```

```python
rede_neural = Sequential([
    tf.keras.layers.InputLayer(shape=(4,)),
    # camada de features input
    tf.keras.layers.Dense(units = 4, activation='relu'),
    # camada ocullta
    tf.keras.layers.Dense(units = 4, activation='relu'),
    tf.keras.layers.Dense(units=3, activation='softmax')
    # sotmax é mais utilizado em multiclasses
])
```

```python
rede_neural.summary()
```

```python
rede_neural.compile(optimizer='adam', loss='categorical_crossentropy', metrics = ['categorical_accuracy'])
# parametros de multiclasse para o modelo
```

```python
rede_neural.fit(x_train,y_train, batch_size= 10, epochs=1000)
# treina o modelo com dados de treino
```

```python
rede_neural.evaluate(x_test, y_test)
# taxa de erro(loss) e accuracy
```

```python
previsoes = rede_neural.predict(x_test)
# previsoes com o modelo
print(previsoes)
```

```python
previsoes = previsoes > 0.5
# troca os numeros por bool para facilitar a visualização
print(previsoes)
```

```python
y_test2 = [np.argmax(t) for t in y_test]
# pega somente os maiores valores, indicando qual classe é(0,1,2)
print(y_test2)
```

```python
previsoes2 = [np.argmax(t) for t in previsoes]
print(previsoes2)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test2, previsoes2)
```

```python
confusion_matrix(y_test2, previsoes2)
```
