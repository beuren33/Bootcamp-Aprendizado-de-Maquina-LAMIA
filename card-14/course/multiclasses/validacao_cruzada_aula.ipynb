```python
import pandas as pd
import tensorflow as tf
import sklearn
import scikeras
```

```python
from scikeras.wrappers import KerasClassifier
from sklearn.preprocessing import LabelEncoder
from tensorflow.keras import utils as np_utils
from tensorflow.keras import backend as k
from tensorflow.keras.models import Sequential
from sklearn.model_selection import cross_val_score
```

```python
base = pd.read_csv('/home/beuren/Documentos/LAMIA-Bootcamp/card-14/data/classificação mais classes/iris.csv')
x = base.iloc[:, 0:4].values
y = base.iloc[:, 4].values
# divide os dados em treino e teste
```

```python
labelencoder = LabelEncoder()
y = labelencoder.fit_transform(y)
y = np_utils.to_categorical(y)
# transforma os dados reais, os rotulos em numeros e depois em catgorias
```

```python
def criar_rede():
    k.clear_session()
    rede_neural = Sequential([
        tf.keras.layers.InputLayer(shape=(4,)),
        tf.keras.layers.Dense(units=4, activation='relu'),
        tf.keras.layers.Dense(units=4, activation='relu'),
        tf.keras.layers.Dense(units=3, activation='softmax')])
    rede_neural.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['categorical_accuracy'])
    return rede_neural
# cria funcao de rede neural com 3 saidas
```

```python
rede_neural = KerasClassifier(model = criar_rede, epochs = 250, batch_size = 10)
```

```python
resultados = cross_val_score(estimator=rede_neural, X = x, y = y, cv = 10, scoring = 'accuracy')
#validacao cruzada vai testar diversas divisoes para a base de dados
```
