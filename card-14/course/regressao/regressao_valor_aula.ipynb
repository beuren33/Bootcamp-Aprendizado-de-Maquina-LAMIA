```python
onehotencoder= ColumnTransformer(transformers=[("OneHot", OneHotEncoder(),[0,1,3,5,8,9,10])], remainder='passthrough')
# transforma as colunas categoricas sem distinguir registros com maior importancia, diferente do labelENcoder
```

```python
x = onehotencoder.fit_transform(x).toarray()
# aplica a categorização
x
```

```python
(316 +1)/2
```

```python
regressor= Sequential([
    tf.keras.layers.InputLayer(shape=(316,)),
    tf.keras.layers.Dense(units=158, activation='relu'),
    tf.keras.layers.Dense(units=158, activation='relu'),
    tf.keras.layers.Dense(units=1, activation='linear'),
])
# cria a rede neural
```

```python
regressor.summary()
```

```python
regressor.compile(loss='mean_absolute_error', optimizer='adam', metrics=['mean_absolute_error'])
# coloca metrica de melhora dos pesos
```

```python
regressor.fit(x,y,batch_size=300, epochs=100)
# treina o modelo
```

```python
x.shape
```

```python
previsoes = regressor.predict(x)
# faz previsoes com os dados
```
