```python
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
sentences=[
    'I like eggs and ham',
    'I love chocolate and bunnies',
    'I hate onions'
]
# lista de frases para testar o empedding
```

```python
MAX_VOCAB_SIZE=20000 # limite de palavras dentro da lista
tokenizer = Tokenizer(num_words=MAX_VOCAB_SIZE)# cria o objeto com limite de palavras
tokenizer.fit_on_texts(sentences)# itera sobre a frase construindo um dcionario de palavras, criando para cada palavra um indice
print(sentences)
sequences = tokenizer.texts_to_sequences(sentences)# transforma cada palavra em inteiros
```

```python
print(sequences)
```

```python
tokenizer.word_index# mostra como ficou a lista com os indicces
```

```python
data=pad_sequences(sequences)
# preenche os numeros que faltam para dar 5 no vetor
data
```

```python
MAX_SEQUENCE_LENGTH = 5
data = pad_sequences(sequences,maxlen=MAX_SEQUENCE_LENGTH)
# mesma coisa da celula anterior mas com uma constante
data
```

```python
data = pad_sequences(sequences,maxlen=MAX_SEQUENCE_LENGTH,padding='post')
# com padding post os zeros vao para o final
data
```

```python
data = pad_sequences(sequences,maxlen=6)
# uma vlor maior que a quantidade de palavras resulta na adição de novos zeros
data
```

```python
data = pad_sequences(sequences,maxlen=4)
# diminui o tamanho de palavras no vetor
data
```

```python
data = pad_sequences(sequences,maxlen=4, truncating='post')
# tira nno os menores numeros
data
```
