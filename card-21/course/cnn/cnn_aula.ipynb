```python
import torch
```

```python
x_train = torch.tensor([[[[4, 5, 6, 7], [5, 6, 7, 8], [8, 9, 10, 11], [4, 6, 7, 8]]], [[[-4, 5, 6, -7], [5, -6, 7, 8], [-8, 9, -10, 11], [-4, -6, -7, -8]]]]).float()
x_train.div_(8)
y_train = torch.tensor([0, 1]).float()
```

```python
print(x_train.shape)
print(y_train.shape)
```

```python
import torch.nn as nn
from torch.optim import Adam

def get_model():
    model = nn.Sequential(
        nn.Conv2d(in_channels=1, out_channels=1, kernel_size=3),
        nn.MaxPool2d(kernel_size=2),
        nn.ReLU(),
        nn.Flatten(),
        nn.Linear(1,1),
        nn.Sigmoid()
    )

    loss_fn = nn.BCELoss()
    optimizer = Adam(model.parameters(), lr=0.01)

    return model, loss_fn, optimizer
```

```python
model, loss, opt = get_model()
```

```python
model
```

```python
def train_batch(x,y,model,opt,loss):
    model.train()
    opt.zero_grad()
    prediction=model(x)
    val_loss = loss(prediction.squeeze(),y.squeeze())
    val_loss.backward()
    opt.step()

    return val_loss.item()
```

```python
from torch.utils.data import DataLoader,Dataset,TensorDataset

train_dl = DataLoader(TensorDataset(x_train,y_train))
```

```python
for epoch in range(100):
    for ix, batch in enumerate(train_dl):
        x,y = batch
        batch_loss = train_batch(x,y,model,opt,loss)
```
