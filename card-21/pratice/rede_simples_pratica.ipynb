```python
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
```

```python
x = torch.randint(1, 100, (1000, 2)).float()#cria numeros aleatorios e 1 a 100 com o shape de 1000 linhas e 1 coluna
y = (x[:,0]**2 + x[:,1]).unsqueeze(1).float()# para o valor target usei umafuncao de nao linear para simular a diferença de um dataset
```

```python
x = torch.tensor(x, dtype=torch.float32)
y = torch.tensor(y, dtype=torch.float32)
```

```python
class Dataset(Dataset):
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __len__(self):
        return len(self.x)

    def __getitem__(self, idx):
        return self.x[idx], self.y[idx]
```

```python
df = Dataset(x,y)
dl = DataLoader(df, batch_size=20, shuffle=True)
```

```python
class RedeNeural(nn.Module):
    def __init__(self):
        super().__init__()
        self.layer1 = nn.Linear(2, 4)
        self.activation = nn.ReLU()
        self.layer2 = nn.Linear(4, 1)


    def forward(self, x):
        x = self.activation(self.layer1(x))# empilhei tudo em uma linha so
        return self.layer2(x)
```

```python
model = RedeNeural()
loss = nn.MSELoss()
optimizer = torch.optim.SGD(model.parameters(), lr=0.01)
```

```python
loss_list=[]
# mesmo padrao a aula
for _ in range(100):
    for data in dl:
        optimizer.zero_grad()
        x, y = data# pega o dado de treino e seu valor real
        l = loss(model(x), y)# compara a previsao do modeo com o valor real
        l.backward()
        optimizer.step()# atualiza os pesos da rede
        loss_list.append(l.item())
```

```python
model.eval()
# faz umma avaliação do modelo com alguns dados de teste

teste = torch.tensor([[5.0, 3.0], [10.0, 2.0]])
with torch.no_grad():  
    pred = model(teste)

print(pred)
```
