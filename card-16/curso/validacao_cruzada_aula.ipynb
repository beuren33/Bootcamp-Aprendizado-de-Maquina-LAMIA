```python
import keras
import numpy as np
import tensorflow as tf
import sklearn
```

```python
from tensorflow.keras.datasets import mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import InputLayer, Dense, Flatten, Conv2D, MaxPooling2D
from tensorflow.keras import utils as np_utils
import numpy as np
from sklearn.model_selection import StratifiedKFold
```

```python
(x, y), (X_teste, y_teste) = mnist.load_data()
x = x.reshape(x.shape[0], 28, 28, 1)
# cloca todos os dados em formato de ideal para a rede neural, tamanho da imagem 28x28 1 entrada de cor que é a escala de cinza
x = x.astype('float32')
# transforma em float para fazer a divisão a abaixo
x /= 255
# normaliza para a escala de 0-1
y = np_utils.to_categorical(y, 10)
# transforma em categoria as classes que serao usadas como saida
```

```python
seed = 5
np.random.seed(seed)
```

```python
kfold = StratifiedKFold(n_splits = 5, shuffle=True, random_state = seed)
# define quantoss splits serao separados a base com base no sseed definido
```

```python
resultados = []
for indice_treinamento, indice_teste in kfold.split(x, np.zeros(shape = (y.shape[0], 1))):
# separa em arrays os dados que vao para o treino e quais vao para o teste, para cada separação separa pelo kfold
  print("Indices treinamento: ", indice_treinamento, "Indice teste: ", indice_teste)
  classificador = Sequential()
  classificador.add(InputLayer(shape=(28, 28, 1)))# camada de entrada de cada pixel
  classificador.add(Conv2D(32, (3, 3), activation='relu'))# faz a matriz de caracteristicas para retornar somente as principais
  classificador.add(MaxPooling2D(pool_size=(2,2)))# diminui a matriz de caracteristicas para 2x2
  classificador.add(Flatten())# transforma a matriz de caracteristicas em um vetor
  classificador.add(Dense(units=128, activation='relu'))# camada oculta
  classificador.add(Dense(units=10, activation='softmax'))# camda de saida com 10 classes
  classificador.compile(loss='categorical_crossentropy', optimizer='adam', metrics=['accuracy'])# define metricas para a rede aprender e otimizar
  classificador.fit(x[indice_treinamento], y[indice_treinamento], batch_size = 128, epochs = 5)# define o treinamento e parametros
  precisao = classificador.evaluate(x[indice_teste], y[indice_teste])# pega a avaliação~de como foi a rede noo teste e adiciona na lista de resultados
  resultados.append(precisao[1])
```
