### Tive que usar um metodo de streaming, entao quando limpava ja escrevia no arquivo, pois nem com a GPU do colab estava dando para escrever no arquivo apos a limpeza

### Como utilizado em outro arquivo de limpeza, usei tecnicas de NLP(REGEX) aprendidas do card de NLP para realizar a limpeza dos dados

```python
import re
from datasets import load_dataset
from google.colab import drive
drive.mount('/content/drive')


dataset = load_dataset("wikimedia/wikipedia", "20231101.pt", split="train", streaming=True)
print(dataset)
print(dataset[0]["text"][:500])
```

```python
def limpeza(text):
    # remove marcadores de referência
    text = re.sub(r"\[\d+\]", "", text)
    # remove espaços repetidos
    text = re.sub(r"[ \t]+", " ", text)
    # tira o espaço entre paragrafos para ate 2 espaços
    text = re.sub(r"\n{3,}", "\n\n", text)
    # remove espaços no fim de cada linha
    text = re.sub(r"[ \t]+\n", "\n", text)
    return text.strip()


def valido(text):
    # passa somente artigos com no minimo 200 caracateres
    return len(text) >= 200
```

```python
ct = 0

with open("/content/drive/MyDrive/LAMIA - Bootcamp/card27/llm/data.txt", "w", encoding="utf-8") as f:
    for row in dataset:
        # para cada linha no dataset chama a função de limpeza
        txt = limpeza(row["text"])
        if valido(txt):
            # escreve direto no arquivo
            f.write(txt)
            f.write("\n\n")
            ct += 1
print(f"total depois: {ct}")
```
