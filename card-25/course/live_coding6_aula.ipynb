```python
import spacy

text = "This is a sample number (555) 555-5555."
# define um texto

nlp = spacy.blank("en")
# cria pipeline vazio
ruler = nlp.add_pipe("entity_ruler")
# adiciona uma uma rule ao pipeline

patterns = [
    {
        "label": "PHONE_NUMBER",
        "pattern": [{"TEXT": {"REGEX": "((\d){3}-(\d){4})"}}]
    }
]
# regex procura pela estrutura e nao depoende dos caracteres especificos
#aplica o regex xom 3 numeros separado por um - depois quatro numeros
# no doc é considerado 3 tokens
ruler.add_patterns(patterns)
# adciona a regra ao pipeline

doc = nlp(text)
for ent in doc.ents:
    print(ent.text, ent.label_) 
```

```python
text = "This is a sample number 5555555."

nlp = spacy.blank("en")
ruler = nlp.add_pipe("entity_ruler")

patterns = [
    {
        "label": "PHONE_NUMBER",
        "pattern": [{"TEXT": {"REGEX": "((\d){5})"}}]
    }
]
# mesma coisa so que agora procurando a um unico token
ruler.add_patterns(patterns)

doc = nlp(text)
for ent in doc.ents:
    print(ent.text, ent.label_)
```
