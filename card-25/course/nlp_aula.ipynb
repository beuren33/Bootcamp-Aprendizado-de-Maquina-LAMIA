```python
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
Sequencias = [
    'I love my cat',
    'I love my dog',
    'You love my dog!',
    'Do you think my dog is amazing?'
]
# sequencia de frases que vao ser tokenizadas, transformadas em indices para a representação de cada uma
```

```python
tokenizer = Tokenizer(num_words=1000, oov_token="<OOV>")
#instacia o objeto com ummm numero limite de palavras conhecidas, e uma palavra que reresenta as desconhecidas
tokenizer.fit_on_texts(Sequencias)
# transforma as palavras em indices, assim criando um vocabulario
word_index = tokenizer.word_index
# exibe o vocabulario criado
```

```python
sequences = tokenizer.texts_to_sequences(Sequencias)
# pega as frases e transforma em sequencias de indices, de acordo com o vocabulario criado

pad_sequences = pad_sequences(sequences, padding='post', maxlen=5, truncating='post')
# aplica um padding as sequencias que nao tem um determinado numero de indices, r exemplo se a maior sequncia tiver 10 indices e uma sequencia tiver 5 indices ela vai receber 5 zeros para que todas tenham o mesmo tamanho
print(pad_sequences)
print(word_index)
print(sequences)
```
