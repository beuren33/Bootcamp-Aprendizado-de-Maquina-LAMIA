```python
import spacy
```

```python
nlp = spacy.load("en_core_web_sm")
# carrega modelo
text = "West Chestertenfieldville was referenced in Mr. Deeds."
```

```python
doc = nlp(text)
# cria um doc com o texto anterior
```

```python
for ent in doc.ents:
    print (ent.text, ent.label_)
    # entidades encontradas no texto baseado em Machine Learning
```

```python
ruler = nlp.add_pipe("entity_ruler")
# adiciona ao pipeline um entity_ruler 
```

```python
nlp.analyze_pipes()
# mostrando o  pipeline
```

```python
patterns = [
        {"label": "GPE", "pattern": "West Chestertenfieldville"}
]
# definindo as regras que serao aplicadas ao texto
```

```python
ruler.add_patterns(patterns)
# adiciona a regra para a rule do pipeline
```

```python
doc2 = nlp(text)
for ent in doc2.ents:
    print (ent.text, ent.label_)
    # segunda tentatva após aplicar a regra
    # aqui nao fez diferença por que a regra foi adicionada ao final do pipeline e o NER ja havia identificado as entidades
```

```python
nlp2 = spacy.load("en_core_web_sm")
# cria um pipeline sem nada
```

```python
ruler = nlp2.add_pipe("entity_ruler", before="ner")
# adiciona a regra antes do NER
```

```python
ruler.add_patterns(patterns)
```

```python
doc = nlp2(text)
```

```python
for ent in doc.ents:
    print (ent.text, ent.label_)
    # agora ao aplicar a regra podemos ver que mudou porem não pegoou o Mr.Deeds
```

```python
nlp2.analyze_pipes()
```

```python
nlp3 = spacy.load("en_core_web_sm")
# cria mais um pipeline
```

```python
ruler = nlp3.add_pipe("entity_ruler", before="ner")
# adiciona o rule antes do NER
```

```python
patterns = [
        {"label": "GPE", "pattern": "West Chestertenfieldville"},
        {"label": "FILM", "pattern": "Mr. Deeds"}
]
# mudança de regras paara o Mr. Deeds
```

```python
ruler.add_patterns(patterns)
```

```python
doc = nlp3(text)
```

```python
for ent in doc.ents:
    print (ent.text, ent.label_)
    # agora pegou corretamente a entidade
```
