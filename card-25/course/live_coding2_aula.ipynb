```python
import numpy as np

your_word = "country"
# palavra que vai ser procuurad

ms = nlp.vocab.vectors.most_similar(np.asarray(
    [nlp.vocab.vectors[nlp.vocab.strings[your_word]]]# busca o indice que corresponde a palavra country no banco vetorial
    ),n=10)# pega as 10 palavras mais proximas
words = [nlp.vocab.strings[w] for w in  ms[0][0]]# converte os indices numericos em palavras
distances=ms[2]
print(words)
```

```python
doc1 = nlp("I like salty fries and hamburgers.")
doc2 = nlp("Fast food tastes very good.")
# define dois novos docs
```

```python
print(doc1, "<->", doc2, doc1.similarity(doc2))
# essa funcao similarity calcula o quao as dois docs se igualam semânticamente
```

```python
doc3 = nlp("The Empire State Building is in New York.")
```

```python
print(doc1, "<->", doc3, doc1.similarity(doc3))
# compara agora com um doc que nao possui tanta semântica
```

```python
doc4 = nlp("I enjoy oranges.")
doc5 = nlp("I enjoy apples.")
# docs com frases quase iguais
```

```python
print(doc4, "<->", doc5, doc4.similarity(doc5))
# similaridade muito alta
```

```python
doc6 = nlp("I enjoy burgers.")
# doc que so muda de fruta para hamburguer
```

```python
print(doc4, "<->", doc6, doc4.similarity(doc6))
# comparando o doc4 de laranjas com hambuguers do doc6
# resultado alto, mostrando que importa a semantica e nao totalmente as mesmas palavras
```

```python
french_fries = doc1[2:4]
burgers = doc1[5]
print(french_fries, "<->", burgers, french_fries.similarity(burgers))
# agora a comparação é feita por um span do doc1 e um token de hamburguers
# aplicando o mesmo metodo de similaridade
```
