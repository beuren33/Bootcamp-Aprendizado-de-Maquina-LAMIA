```python
with open("/home/beuren/Documentos/Bootcamp-LAMIA/card-25/data/wiki_us.txt", "r") as file:
    text = file.read()
    # le arquivo de texxto
```

```python
print(text)
```

```python
doc = nlp(text)
# cria um objeto doc com o texto carregado
```

```python
print(len(text))# quantidade de caracteres do texto bruto
print(len(doc))# quantidade de tokens no doc
```

```python
for token in text[:10]:
    print(token)
    # printa cada caracter do texto bruto
```

```python
for token in doc[:10]:
    print(token)
    # printa cada token que esta no doc
```

```python
for token in text.split()[:10]:
    print(token)
    # usando o split normal
    # o split nao reconhece que abreviações como U.S.A dao sentido a uma unica entidade
```

```python
doc2 = nlp("Mike and Joy's plane football")
# cria outro objeto doc
```

```python
print(doc2)
```

```python
for token in doc2:
    print(token.text, token.pos_,token.dep_)
    # para cada token imprime o texto a classe gramatical e duncao sintática
```
